```python
import pandas as pd
import requests
from io import StringIO

# Define the URL
url = "https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/2244,1983,702,6322,5957/property/MolecularWeight,LogP,InChIKey,CanonicalSMILES/CSV"

# Send GET request to the URL
response = requests.get(url)

# Check if request was successful
if response.status_code == 200:
    data = StringIO(response.text)         # Convert response text into a file-like object
    df = pd.read_csv(data)                 # Read into pandas DataFrame
    print("✅ Data loaded successfully!\n")
    print(df.head())                       # Display first few rows
else:
    print("❌ Failed to fetch data from PubChem.")
    print("HTTP Status Code:", response.status_code)
```

```python
print(df.columns)
print(df.dtypes)
print(df.isnull().sum())
```

```python
# Rename for clarity (optional)
df.rename(columns={"CanonicalSMILES": "SMILES"}, inplace=True)

# Drop missing values if any
df.dropna(inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot 1: Molecular Weight vs LogP
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="MolecularWeight", y="LogP")
plt.title("Molecular Weight vs LogP")
plt.xlabel("Molecular Weight")
plt.ylabel("LogP")
plt.grid(True)
plt.show()

# Plot 2: LogP vs Compound (using InChIKey as ID)
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="LogP", y="InChIKey")
plt.title("LogP Values Across Compounds")
plt.xlabel("LogP")
plt.ylabel("Compound (InChIKey)")
plt.grid(True)
plt.show()
```

**FULL CODE**

```python
# Step 1: Load Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load Dataset
url = "https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/2244,1983,702,6322,5957/property/MolecularWeight,LogP,InChIKey,CanonicalSMILES/CSV"
df = pd.read_csv(url)

# Step 3: Inspect Dataset
print("Column Names:", df.columns.tolist())
print("\nData Types:\n", df.dtypes)
print("\nMissing Values:\n", df.isnull().sum())
print("\nBasic Info:")
df.info()
print("\nFirst few rows:")
print(df.head())

# Step 4: Basic Cleaning (optional renaming for clarity)
df.rename(columns={
    "CID": "CompoundID",
    "InChIKey": "CompoundKey",
    "CanonicalSMILES": "SMILES"
}, inplace=True)

# Optional: Convert categorical ID to string
df["CompoundID"] = df["CompoundID"].astype(str)

# Step 5: Summary Statistics
print("\nSummary Statistics:\n", df.describe())

# Step 6: Visualize Molecular Weight vs LogP
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="MolecularWeight", y="LogP", hue="CompoundID", palette="viridis", s=100)
plt.title("Molecular Weight vs LogP", fontsize=14)
plt.xlabel("Molecular Weight (g/mol)")
plt.ylabel("LogP")
plt.grid(True)
plt.legend(title="Compound ID")
plt.tight_layout()
plt.show()

# Step 7: Visualize LogP per Compound
plt.figure(figsize=(9, 5))
sns.barplot(data=df, x="CompoundKey", y="LogP", palette="coolwarm")
plt.title("LogP Values Across Compounds", fontsize=14)
plt.xlabel("Compound (InChIKey)")
plt.ylabel("LogP")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y')
plt.show()

# Step 8: Additional Plot (Optional) - Molecular Weight Distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["MolecularWeight"], kde=True, bins=10, color='teal')
plt.title("Distribution of Molecular Weights", fontsize=14)
plt.xlabel("Molecular Weight (g/mol)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.grid(axis='y')
plt.show()

# Optional Step: Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(df[["MolecularWeight", "LogP"]].corr(), annot=True, cmap="mako", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
